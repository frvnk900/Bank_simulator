```python
 def change_pin(account:dict,user_name:str):
    
     
    if user_name in account:
        
        current_pin=input("\nEnter Your current pin: ")
        if current_pin == account[user_name]["pin"]:
             new_pin=input("\nEnter Your New pin: ")
             account[user_name]["pin"]=new_pin
             print("===PIN CHANGED SUCCESSFUL===")
        else:
           print("\n==INCORRECT PIN!==")
    else:
         print("\n==NO such user found==")
```
